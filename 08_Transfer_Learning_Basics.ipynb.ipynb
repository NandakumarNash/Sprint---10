## 1. What is Transfer Learning?

**Transfer learning** means using a model that has already been trained on a large dataset and using its learned knowledge for a new task.

Instead of training a neural network from the beginning, we start with a **pretrained model** and adapt it to our problem.

### Simple Example

Suppose a CNN was already trained on millions of images.

It has learned to recognize things like:

- Edges
- Shapes
- Textures
- Patterns
- Objects

We can reuse this knowledge for a new image classification problem.

### Basic Idea

**Train from scratch:**

`Random Model → Train with Our Data → Final Model`

**Transfer learning:**

`Pretrained Model → Adapt to Our Data → Final Model`

Transfer learning is especially useful when we have a **small or medium-sized dataset**.

### Why do we use it?

Training a deep neural network from scratch can require:

- A large dataset
- More training time
- More computational resources

Transfer learning allows us to **reuse knowledge learned by an existing model**.

### Real-World Example

A pretrained image model already knows how to identify basic visual patterns.

If our new task is to classify **different types of flowers**, we can reuse the pretrained model and train it to recognize the flower categories.

### Important Point

**Transfer learning = Reusing knowledge from a pretrained model for a new task.**

````python
import torch
import torchvision.models as models

# Load a pretrained model
model = models.resnet18(weights="DEFAULT")

print(model)
````

### Simple Concept

A **pretrained model** is a model that has already learned from a large dataset.

Instead of starting with random weights, we start with the knowledge the model has already learned.

### Why are pretrained models useful?

They help us:

- **Save training time** – we don't need to train everything from the beginning.
- **Need less data** – useful when our dataset is small.
- **Reuse learned features** – the model already understands basic patterns such as edges, shapes, and textures.
- **Reduce computational requirements** – training from scratch can require more resources.



### What happens here?

We are **not training the model yet**.

We are simply taking an already-trained model and preparing to reuse its learned knowledge for our new task.

### Real-World Example

Imagine an employee who already knows how to identify different types of objects.

If you give that employee a new but related classification task, they don't need to learn everything from zero.

Similarly, a pretrained model already has useful knowledge that we can reuse.

### Important Point

**Pretrained models are useful because they provide a strong starting point and can reduce the amount of data, time, and computation needed for training.**

## 3. Feature Extraction

### Simple Concept

**Feature extraction** means using the pretrained model to extract useful patterns from our new images.

A pretrained CNN has already learned features such as:

- Edges
- Shapes
- Textures
- Patterns

We can use these learned features without training the entire model again.

### How it works

```text
Input Image
     ↓
Pretrained CNN
     ↓
Learned Features
     ↓
New Classifier
     ↓
Prediction
```

For example, if we are classifying **cats and dogs**, the pretrained CNN can extract features such as ears, fur, shapes, and other visual patterns.

The new classifier uses these extracted features to decide whether the image is a **cat or dog**.


### Important Point

**Feature extraction = Reusing the pretrained model's learned features and using them for a new classification task.**

````python

### Python Example

import torch
import torchvision.models as models

# Load pretrained model
model = models.resnet18(weights="DEFAULT")

# Remove the final classifier
feature_extractor = torch.nn.Sequential(
    *list(model.children())[:-1]
)

print(feature_extractor)
````

````python
```

### Python Explanation

`models.resnet18(weights="DEFAULT")`  
→ Loads ResNet18 with pretrained knowledge.

`model.children()`  
→ Gets the different layers of the ResNet model.

`[:-1]`  
→ Removes the final classification layer.

`torch.nn.Sequential(...)`  
→ Creates a new model using the remaining layers.

Now the model acts mainly as a **feature extractor**.
````

## 4. Frozen Layers

### Simple Concept

A **frozen layer** is a layer whose weights are **not updated during training**.

In transfer learning, we often freeze the pretrained model's backbone because it has already learned useful features.

```text
Pretrained Model
       ↓
Freeze Backbone
       ↓
Keep learned features unchanged
       ↓
Train new classifier
```

### Why freeze layers?

Freezing helps us:

- Keep the pretrained knowledge
- Reduce the number of parameters that need training
- Make training faster
- Reduce the chance of overfitting when the dataset is small


### Important Point

**Frozen layer = Its learned weights remain unchanged during training.**

In a typical transfer-learning setup:

**Pretrained Backbone → Frozen**

**New Classifier → Trainable**

````python

### Python Example

import torch
import torchvision.models as models

# Load pretrained model
model = models.resnet18(weights="DEFAULT")

# Freeze all layers
for parameter in model.parameters():
    parameter.requires_grad = False

print("All pretrained layers are frozen")
````

```

### Python Explanation

`model.parameters()`  
→ Gets all trainable parameters of the model.

`parameter.requires_grad = False`  
→ Tells PyTorch **not to calculate gradients for these parameters**.

Because gradients are not calculated, these weights will **not be updated during training**.

## 5. Trainable Layers

### Simple Concept

A **trainable layer** is a layer whose weights **can be updated during training**.

In transfer learning, we usually keep most of the pretrained backbone frozen and make the **new classifier trainable**.

```text
Pretrained Backbone → Frozen
                         ↓
                  Extract Features
                         ↓
New Classifier → Trainable
                         ↓
                    Prediction
```

### Why make layers trainable?

The new classifier needs to learn how to use the pretrained features for **our specific task**.

For example, a pretrained model may already understand general image features, but our new task may be:

**Image → Cat / Dog**

The new classifier needs to learn this specific classification.


### Important Point

**Trainable layer = A layer whose weights can be updated during training.**

In this example:

**Backbone → Frozen**

**New Classifier → Trainable**

````python

### Python Example

import torch
import torchvision.models as models

# Load pretrained model
model = models.resnet18(weights="DEFAULT")

# Freeze the pretrained layers
for parameter in model.parameters():
    parameter.requires_grad = False

# Replace the final classifier
model.fc = torch.nn.Linear(model.fc.in_features, 2)

# Check trainable parameters
for name, parameter in model.named_parameters():
    if parameter.requires_grad:
        print(name, "is trainable")
````

```

### Python Explanation

`requires_grad = False`  
→ Freezes the pretrained layers.

`model.fc.in_features`  
→ Gets the number of inputs expected by the original classifier.

`torch.nn.Linear(..., 2)`  
→ Creates a new classifier with **2 output classes**.

`requires_grad` for the new classifier is `True` by default.  
→ Therefore, the new classifier can learn during training.

## 6. Fine-Tuning

### Simple Concept

**Fine-tuning** means taking a pretrained model and **unfreezing some of its layers** so they can learn from our new dataset.

Instead of training only the new classifier, we allow selected pretrained layers to update their weights.

```text
Pretrained Model
       ↓
Freeze Most Layers
       ↓
Unfreeze Selected Layers
       ↓
Train on New Dataset
       ↓
Fine-Tuned Model
```

### Why do we use Fine-Tuning?

The pretrained model has already learned general features.

But our new dataset may have specific patterns that the original model did not learn.

Fine-tuning allows selected layers to **adapt to these new patterns**.


### Feature Extraction vs Fine-Tuning

**Feature Extraction:**

```text
Pretrained Model
      ↓
All Backbone Layers Frozen
      ↓
New Classifier Trained
```

**Fine-Tuning:**

```text
Pretrained Model
      ↓
Some Backbone Layers Unfrozen
      ↓
Selected Layers + Classifier Trained
```

### Important Point

**Fine-tuning = Unfreezing selected pretrained layers and training them on the new dataset.**

````python

### Python Example

import torch
import torchvision.models as models

# Load pretrained model
model = models.resnet18(weights="DEFAULT")

# Freeze all layers
for parameter in model.parameters():
    parameter.requires_grad = False

# Unfreeze the last layer
for parameter in model.layer4.parameters():
    parameter.requires_grad = True

# Replace classifier
model.fc = torch.nn.Linear(model.fc.in_features, 2)

print("Selected layers are trainable")
````

```

### Python Explanation

`requires_grad = False`  
→ Initially freezes all pretrained layers.

`model.layer4.parameters()`  
→ Selects the last major convolutional block of ResNet18.

`requires_grad = True`  
→ Makes this selected block trainable.

`model.fc = ...`  
→ Replaces the original classifier with a new classifier for **2 classes**.

Now:

```text
Early Layers → Frozen
Last CNN Block → Trainable
New Classifier → Trainable
```

## 7. Domain Similarity

### Simple Concept

**Domain similarity** means how similar the pretrained model's original data is to our new data.

If the two datasets are similar, transfer learning usually works well because the pretrained model has already learned useful features for the new task.

### Example

Suppose a model was pretrained on **general images**.

Our new task is also **image classification**.

```text
Pretrained Domain          New Domain

General Images     →       Animal Images
       ↓                         ↓
     Similar                 Related
```

The pretrained model's learned features can be useful for the new task.

### Why is Domain Similarity Important?

When the domains are similar:

- Learned features can be reused effectively.
- Less training may be required.
- Feature extraction can work well.
- Fine-tuning can further adapt the model.


### Important Point

**Domain similarity = Similarity between the data used for pretraining and the data used for the new task.**

**More similarity → pretrained features are more likely to be useful.**

````python


import torch
import torchvision.models as models

# Pretrained model
model = models.resnet18(weights="DEFAULT")

# Freeze pretrained layers
for parameter in model.parameters():
    parameter.requires_grad = False

# Replace classifier for our new task
model.fc = torch.nn.Linear(model.fc.in_features, 3)

print("Model adapted for 3 classes")
````

```

### Python Explanation

`resnet18(weights="DEFAULT")`  
→ Loads a model that has already learned from a large image dataset.

`requires_grad = False`  
→ Keeps the pretrained features unchanged.

`model.fc = torch.nn.Linear(..., 3)`  
→ Replaces the original classifier with one that predicts **3 new classes**.

Because both the original and new tasks involve images, the pretrained visual features can be useful.

## 8. Domain Mismatch

### Simple Concept

**Domain mismatch** happens when the data used to train the pretrained model is very different from our new dataset.

For example:

```text
Pretrained Model
       ↓
Natural Images
       ↓
New Task
       ↓
Medical X-ray Images
```

The two domains are very different.

The pretrained model may still provide some useful basic features, but its learned features may not match the new task very well.

### What happens with Domain Mismatch?

When there is a large difference between the domains:

- Feature extraction may not work as well.
- More fine-tuning may be required.
- More layers may need to be unfrozen.
- A larger new dataset may be helpful.


### Important Point

**Domain mismatch = The pretrained data and new data are very different.**

When domain mismatch is high, **fine-tuning is often more useful than using the pretrained model only as a fixed feature extractor.**

````python

### Python Example

import torch
import torchvision.models as models

# Load pretrained model
model = models.resnet18(weights="DEFAULT")

# Freeze all layers first
for parameter in model.parameters():
    parameter.requires_grad = False

# Unfreeze the last block for adaptation
for parameter in model.layer4.parameters():
    parameter.requires_grad = True

# Replace classifier
model.fc = torch.nn.Linear(model.fc.in_features, 2)

print("Model is ready for fine-tuning")
````

```

### Python Explanation

`requires_grad = False`  
→ Freezes the pretrained layers.

`model.layer4.parameters()`  
→ Selects the last major feature-learning block.

`requires_grad = True`  
→ Allows this block to learn new features from our dataset.

`model.fc = ...`  
→ Replaces the old classifier with a new 2-class classifier.

This allows the model to **adapt some of its pretrained knowledge** to the new domain.

## Transfer Learning Workflow

### Workflow

**Pretrained Model → Freeze Backbone → Replace Classifier → Train**

This is the basic **feature extraction** approach.

### Step 1: Pretrained Model

We start with a model that has already learned useful image features.

```text
Pretrained Model
       ↓
Already learned features
```

### Step 2: Freeze Backbone

We freeze the pretrained layers so their weights do not change.

```text
Backbone → Frozen
```

The model continues to use the features it already learned.

### Step 3: Replace Classifier

The original classifier is replaced with a new classifier for our classes.

For example:

```text
Original Model → 1000 classes
Our Task       → 2 classes
```

So we replace the classifier with one that produces **2 outputs**.

### Step 4: Train

Only the new classifier is trained.

```text
Pretrained Backbone → Frozen
                         ↓
                    Features
                         ↓
New Classifier → Trainable
                         ↓
                    Prediction
```


### Final Structure

```text
Pretrained ResNet
       ↓
Frozen Backbone
       ↓
Feature Extraction
       ↓
New 2-Class Classifier
       ↓
Training
```

### Important Point

In this approach, **we reuse the pretrained features and train only the new classifier**.

````python

### Python Example

import torch
import torchvision.models as models

# 1. Load pretrained model
model = models.resnet18(weights="DEFAULT")

# 2. Freeze the backbone
for parameter in model.parameters():
    parameter.requires_grad = False

# 3. Replace the classifier
model.fc = torch.nn.Linear(model.fc.in_features, 2)

# 4. Check trainable parameters
for name, parameter in model.named_parameters():
    if parameter.requires_grad:
        print(name, "is trainable")
````

```

### Python Explanation

`models.resnet18(weights="DEFAULT")`  
→ Loads the pretrained ResNet18 model.

`requires_grad = False`  
→ Freezes the pretrained model parameters.

`model.fc = torch.nn.Linear(...)`  
→ Replaces the original classifier with a new classifier.

`2`  
→ Means our new task has **2 classes**.

`requires_grad` for the new classifier is `True` by default.  
→ Therefore, the classifier can learn during training.
