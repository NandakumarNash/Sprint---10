### Simple Concept

**Internal distribution changes** means the values going into a neural network layer keep changing during training.

For example:

```text
Input
  ↓

  ↓
Activation
  ↓
Next Layer
```

When earlier layers update their weights, the output values they produce also change.

So, the next layer keeps receiving data with a changing distribution.

### Why is this a problem?

The next layer has to continuously adjust to these changing values.

This can make neural network training:

- Less stable
- Slower
- More difficult to optimize

### Why Batch Normalization is used

**Batch Normalization helps keep the values more stable during training.**

It normalizes the values within a batch before passing them to the next layer.

```text
CNN Layer
   ↓
Batch Normalization
   ↓
Activation
   ↓
Next Layer
```

### Simple Concept

Both **Normalization** and **Standardization** are methods used to change the scale of data.

### Normalization

Normalization usually converts values into a fixed range, commonly **0 to 1**.

Example:

```text
Original values:  0, 50, 100
Normalized:       0, 0.5, 1
```

It is useful when we want values to be within a specific range.

### Standardization

Standardization changes the data so that it has approximately:

- Mean = 0
- Standard deviation = 1

For example, a value can be converted into a **z-score** based on how far it is from the mean.



### Difference

| Normalization | Standardization |
|---|---|
| Usually scales to a fixed range | Centers data around 0 |
| Common range: 0 to 1 | Mean ≈ 0 |
| Useful when a bounded range is needed | Useful when data needs to be centered and scaled |

### In this CNN topic

**Batch Normalization is different from simply preprocessing your input data.**

BatchNorm operates on values **inside the neural network during training**, helping keep those intermediate values more stable.

### Simple Concept

**Dropout** is a regularization technique used to reduce **overfitting** in a neural network.

During training, Dropout randomly turns off some neurons.

```text
Before Dropout

Neuron → Neuron → Neuron → Neuron → Neuron

              ↓ Dropout

Neuron → OFF   → Neuron → OFF   → Neuron
```

The neurons that are turned off are selected randomly during each training step.

### Why do we need Dropout?

A neural network can sometimes **memorize the training data** instead of learning useful patterns.

Dropout prevents the model from depending too much on particular neurons.

This can help the model **generalize better to unseen data**.

### Example

If we use:

```python
nn.Dropout(0.5)
```

approximately **50% of the neurons** in that layer are randomly dropped during training.

### Important

Dropout is active during **training**.

During **evaluation/testing**, Dropout is turned off so that all neurons are used.

```text
Training  → Dropout ON
Testing   → Dropout OFF
```

### Simple Concept

A neural network behaves slightly differently during **training** and **evaluation**.

### Training Mode

During training:

- The model learns from the training data.
- Weights are updated.
- Dropout is active.
- Batch Normalization uses the current batch statistics.

```text
Training
   ↓
Learn patterns
   ↓
Update weights
   ↓
Dropout ON
```

### Evaluation Mode

During evaluation/testing:

- The model does not update its weights.
- Dropout is turned off.
- Batch Normalization uses its learned running statistics.

```text
Evaluation
   ↓
Make predictions
   ↓
Weights NOT updated
   ↓
Dropout OFF
```

### Why is this important?

If we use the wrong mode, the model may behave differently when making predictions.

So we use:

```python
model.train()
```

for training and:

```python
model.eval()
```

for evaluation/testing.

### Simple Concept

`model.train()` puts the neural network into **training mode**.

We use it when the model is learning from training data.

```python
model.train()
```

### What happens?

When training mode is enabled:

- The model learns and updates its weights.
- **Dropout is ON.**
- **Batch Normalization uses the current training batch statistics.**

Example:

```python
model.train()

output = model(X_train)
```

Here, the model processes the training data in **training mode**.

### Important

`model.train()` does **not** itself update the weights.

The weights are updated when we perform the training steps such as:

```python
loss.backward()
optimizer.step()
```

So:

```text
model.train()
      ↓
Training mode ON
      ↓
Forward pass
      ↓
Calculate loss
      ↓
Backward pass
      ↓
Update weights
```

### Simple Concept

`model.eval()` puts the neural network into **evaluation mode**.

We use it when we want to evaluate the model or make predictions.

```python
model.eval()
```

### What happens?

When evaluation mode is enabled:

- Model weights are **not updated** by the evaluation process.
- **Dropout is OFF.**
- **Batch Normalization uses its learned running statistics.**

Example:

```python
model.eval()

output = model(X_test)
```

### Important

`model.eval()` does **not** calculate accuracy or loss by itself.

It only changes the model from **training mode → evaluation mode**.

For example:

```python
model.eval()

with torch.no_grad():
    output = model(X_test)
```

`torch.no_grad()` prevents gradient calculation during evaluation.

### Simple Difference

```text
model.train()
→ Training mode
→ Dropout ON
→ BatchNorm uses batch statistics

model.eval()
→ Evaluation mode
→ Dropout OFF
→ BatchNorm uses running statistics
```
